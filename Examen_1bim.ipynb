1. **Líneas, tickets y unidades.** ¿Cuántas líneas de venta, cuántos tickets distintos y cuántas unidades vendidas hay en el archivo? Explique qué representa cada conteo.



```python
lineVen=len(ventas)
ticketsDispo=ventas["id_ticket"].nunique()
cantVenta=0
for i in ventas["cantidad"]:
    cantVenta+=i
print("Hay esta cantidad de lines: ",lineVen)
print("Hay esta cantidad de tickets distintos: ",ticketsDispo)
print("Hay esta cantidad de unidades vendidas: ",cantVenta)
```

2. **Tickets por sucursal.** ¿Cuántos tickets se registraron en cada sucursal? ¿Cuál registró más?



```python
contTSur=0
contTNor=0
contTCen=0

for surcursal in ventas["sucursal"]:
    if surcursal=="Sur":
        contTSur+=1
    elif surcursal=="Norte":
        contTNor+=1
    else:
        contTCen+=1

print("Se vendio en el Sur esta cantidad: ",contTSur)
print("Se vendio en el Norte esta cantidad: ",contTNor)
print("Se vendio en el Centro esta cantidad: ",contTCen)

if contTNor>contTSur and contTNor>contTSur:
    print ("Sucursal Norte vendió más")
elif contTSur>contTNor and contTSur>contTCen:
    print("Sucursal Sur vendió más")
else:
    print("Sucursal central vendió más")
```

3. **Métodos de pago.** ¿Cuántos tickets se pagaron con cada método y cuántos tienen ese dato sin registrar?



```python

contEfectivo=0
contTarjeta=0
contTransfe=0
contSi=0

for metodo in ventas["metodo_pago"]:
    if metodo=="Transferencia":
        contTransfe+=1
    elif metodo=="Tarjeta":
        contTarjeta+=1
    elif metodo=="Efectivo":
        contEfectivo+=1
    else:
        contSi+=1

print ("Con el metodo transferencia se tiene: ",contTransfe)
print ("Con el metodo Efectivo se tiene: ",contEfectivo)
print ("Con el metodo Tarjeta se tiene: ",contTarjeta)
print ("Sin metodo se tiene: ",contSi)

```

```python
precioEcri=[]
precioArchi=[]
precioAccesorios=[]
precioCuadern=[]

for categoria,precio in zip(productos["categoria"],productos["precio_lista"]):
    if categoria=="Escritura":
        precioEcri.append(precio)
    elif categoria=="Archivo":
        precioArchi.append(precio)
    elif categoria=="Accesorios":
        precioAccesorios.append(precio)
    else:
        precioCuadern.append(precio)

promE=sum(precioEcri)/len(precioEcri)
minE=min(precioEcri)
maxE=max(precioEcri)

promAR=sum(precioArchi)/len(precioArchi)
minAr=min(precioArchi)
maxAr=max(precioArchi)

promAC=sum(precioAccesorios)/len(precioAccesorios)
minAC=min(precioAccesorios)
maxAC=max(precioAccesorios)

promC=sum(precioCuadern)/len(precioCuadern)
minC=min(precioCuadern)
maxC=max(precioCuadern)

print ("En la categoria Escritura, tiene de maximo: ",maxE,",min: ",minE, " y prmedio: ",promE)
print ("En la categoria Archivo, tiene de maximo: ",maxAr,",min: ",minAr, " y prmedio: ",promAR)
print ("En la categoria Accesorio, tiene de maximo: ",maxAC,",min: ",minAC, " y prmedio: ",promAC)
print ("En la categoria Cuaderno, tiene de maximo: ",maxC,",min: ",minC, " y prmedio: ",promC)
```
